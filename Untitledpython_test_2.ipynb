```python
#emplyee salary analyzer
def calculate_salary(basic, department, rating):
    allowances = {
        "IT": 0.20,
        "HR": 0.15,
        "Sales": 0.25,
        "Finance": 0.18
    }

    performance_bonus = {
        1: 0.02,
        2: 0.05,
        3: 0.08,
        4: 0.12,
        5: 0.15
    }

    allowance = basic * allowances.get(department, 0.10)
    bonus = basic * performance_bonus.get(rating, 0)

    return basic + allowance + bonus


employees = [
    {"name": "Aman", "department": "IT", "basic": 50000, "rating": 5},
    {"name": "Riya", "department": "HR", "basic": 45000, "rating": 4},
    {"name": "Rahul", "department": "Sales", "basic": 40000, "rating": 3},
    {"name": "Neha", "department": "Finance", "basic": 55000, "rating": 4}
]

department_salary = {}

for emp in employees:
    emp["final_salary"] = calculate_salary(
        emp["basic"],
        emp["department"],
        emp["rating"]
    )

    department_salary.setdefault(emp["department"], [])
    department_salary[emp["department"]].append(emp["final_salary"])


highest = max(employees, key=lambda x: x["final_salary"])

print("Employee with Highest Final Salary:")
print(highest["name"], "₹", highest["final_salary"])

print("\nDepartment-wise Average Salary:")
for dept, salaries in department_salary.items():
    average = sum(salaries) / len(salaries)
    print(dept, "₹", round(average, 2))
```
